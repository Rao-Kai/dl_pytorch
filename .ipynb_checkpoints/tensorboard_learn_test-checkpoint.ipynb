```python
from torch.utils.tensorboard import SummaryWriter
from PIL import Image
import numpy as np
```

```python
writer = SummaryWriter("logs")
```

```python
for i in range(100):
    writer.add_scalar("y=2x", 2*i, i) #writer.add_scalar(图表名，y轴值，x轴值)

writer.close()
```

```python
image_path1 = "dataset//train//ants//0013035.jpg"
image_path2 = "dataset//train//ants//7759525_1363d24e88.jpg"
img1 = Image.open(image_path1)
img2 = np.array(img1)
img3 = Image.open(image_path2)
img4 = np.array(img3)
print(img2.shape) #是HWC型，需要在函数中指定，函数默认的是CHW
writer.add_image("ants1", img2, 1, dataformats="HWC") #writer.add_image(名称，图片对象，步数)
writer.add_image("ants1", img4, 2, dataformats="HWC")
```
