```python
from torchvision import transforms
from torch.utils.tensorboard import SummaryWriter
from PIL import Image 
import cv2
#通过transforms.ToTensor解决两个问题
#1、transforms如何使用； 2、为什么需要tensor数据类型
```

```python
#绝对路径
#img_path = "E:\pytorch_learn\dataset\train\ants\0013035.jpg"
#相对路径
img_path = "dataset/train/ants/0013035.jpg"
img = Image.open(img_path)
print(img)
```

```python
#1、transforms如何使用
tensor_trans = transforms.ToTensor()
tensor_img = tensor_trans(img)
print(tensor_img)
```

```python
# 2、为什么需要tensor数据类型
# Tensor数据类型包含使用神经网络需要的信息，如前向方法、后向方法、梯度等
```

```python
cv_img = cv2.imread(img_path) #opencv读进来的是numpy.ndarray格式
```

```python
writer = SummaryWriter("logs")
writer.add_image("Tensor_img", tensor_img)
writer.close()
```
